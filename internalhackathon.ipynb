```python
x_train = preprocessing.fit_transform(x_train)
x_test = preprocessing.transform(x_test)
```

```python
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import xgboost as xgb
```

```python
models = [
    ('SVM', SVR(kernel='linear')),
    ('Linear Regression', LinearRegression()),
    ('xgboost',xgb.XGBRegressor()),
    ('Decision Tree', DecisionTreeRegressor()),
    ('Random Forest', RandomForestRegressor())
]
```

```python
def mean_sqaure_error(x_train,x_test,y_train,y_test,models):
  evaluate = {}
  for name,model in models:
    model.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    mse = mean_squared_error(y_test,y_pred)
    evaluate[name] = mse
  return evaluate
```

```python
mean_sqaure_error(x_train,x_test,y_train,y_test,models)
```

```python
results = mean_sqaure_error(x_train, x_test, y_train, y_test, models)

# Extract model names and MSE values
model_names = list(results.keys())
mse_values = list(results.values())

# Create a bar plot
plt.figure(figsize=(10, 6))
plt.bar(model_names, mse_values)
plt.xlabel("Regression Models")
plt.ylabel("Mean Squared Error (MSE)")
plt.title("MSE Comparison for Different Regression Algorithms")
plt.show()
```
